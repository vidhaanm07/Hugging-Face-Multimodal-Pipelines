```python
from PIL import Image

image = Image.open("image/test_image.jpg.jpg").convert("RGB")

print("Image loaded successfully!")
```

```python
print(image.size)
```

```python
result = image_pipeline(image)

print(result)
```

```python
best_prediction = result[0]

print("Detected Object:", best_prediction["label"])
print("Confidence:", best_prediction["score"])
```

```python
label = result[0]["label"]

caption = f"The image contains a {label}."

print("Generated Caption:")
print(caption)
```

```python
label = result[0]["label"]
confidence = result[0]["score"]

print("===== IMAGE PIPELINE OUTPUT =====")
print("Detected Object :", label)
print("Confidence      :", f"{confidence:.2%}")
print("Generated Caption:", f"The image contains a {label}.")
```

```python
result = image_pipeline(image, top_k=5)

print("===== IMAGE ANALYSIS =====")

for i, prediction in enumerate(result, 1):
    print(
        f"{i}. {prediction['label']} "
        f"({prediction['score']:.2%})"
    )
```
