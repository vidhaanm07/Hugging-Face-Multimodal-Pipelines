```python
def text_pipeline(text):
    positive_words = {
        "good", "great", "excellent", "amazing",
        "awesome", "love", "happy", "perfect", "best"
    }

    negative_words = {
        "bad", "terrible", "awful", "worst",
        "hate", "sad", "poor", "horrible"
    }

    words = text.lower().replace(".", "").replace(",", "").split()

    positive_score = sum(word in positive_words for word in words)
    negative_score = sum(word in negative_words for word in words)

    if positive_score > negative_score:
        return "POSITIVE"
    elif negative_score > positive_score:
        return "NEGATIVE"
    else:
        return "NEUTRAL"
```

```python
text = "I really love this product. It is excellent!"

result = text_pipeline(text)

print("===== TEXT PIPELINE OUTPUT =====")
print("Input     :", text)
print("Sentiment :", result)
```
