```python
import imageio.v2 as imageio

reader = imageio.get_reader("video/SupportFlow_AI Demo Video.mp4")

print("Video opened successfully!")
```

```python
meta = reader.get_meta_data()

print("FPS:", meta.get("fps"))
print("Duration:", meta.get("duration"))
print("Size:", meta.get("size"))
```

```python
frames = []

for i, frame in enumerate(reader):
    if i % 20 == 0:
        frames.append(frame)

    if len(frames) >= 5:
        break

reader.close()

print("Frames extracted:", len(frames))
```

```python
print("Frames extracted:", len(frames))
print("First frame shape:", frames[0].shape)
```

```python
video_analysis = []

for i, frame in enumerate(frames):
    height, width, channels = frame.shape

    video_analysis.append({
        "frame": i + 1,
        "width": width,
        "height": height,
        "channels": channels
    })

print("===== VIDEO PIPELINE OUTPUT =====")

for item in video_analysis:
    print(
        f"Frame {item['frame']}: "
        f"{item['width']}x{item['height']}, "
        f"{item['channels']} channels"
    )
```

```python
print("===== VIDEO PIPELINE SUMMARY =====")
print("Video:", "SupportFlow_AI Demo Video.mp4")
print("Frames analyzed:", len(frames))
print("Frame resolution:", frames[0].shape[1], "x", frames[0].shape[0])
print("Channels:", frames[0].shape[2])
print("Status: Video processed successfully")
```
